```python
from grakel.datasets import fetch_dataset
import grakel.datasets.base as gb

# Patch MUTAG URL to current TU Dortmund mirror
gb.dataset_metadata["MUTAG"]["link"] = "https://www.chrsmrrs.com/graphkerneldatasets/MUTAG.zip"

MUTAG = fetch_dataset("MUTAG", verbose=False)
G = MUTAG.data
y = MUTAG.target
```

```python
from grakel import GraphKernel
sp_kernel = GraphKernel(kernel="shortest_path")
```

```python
from sklearn.model_selection import train_test_split
G_train, G_test, y_train, y_test = train_test_split(G, y, test_size=0.1, random_state=42)
```

```python
K_train = sp_kernel.fit_transform(G_train)
K_test = sp_kernel.transform(G_test)
```

```python
from sklearn.svm import SVC
clf = SVC(kernel='precomputed')
clf.fit(K_train, y_train)
SVC(kernel='precomputed')

y_pred = clf.predict(K_test)
```

```python
from sklearn.metrics import accuracy_score
print("%2.2f %%" %(round(accuracy_score(y_test, y_pred)*100)))
```
